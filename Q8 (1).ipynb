```python
import spacy

# Load the standard English language model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    # Fallback to download the model if not already present
    import os
    os.system("python -m spacy download en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

def proper_noun_annotator(text: str):
    """
    Parses the input text and extracts tokens tagged as proper nouns (PROPN).
    """
    doc = nlp(text)
    proper_nouns = []

    for token in doc:
        if token.pos_ == "PROPN":
            proper_nouns.append({
                "text": token.text,
                "label": "PROPN",
                "start_char": token.idx,
                "end_char": token.idx + len(token.text)
            })

    return proper_nouns

# --- Example Usage ---
sample_text = "Alice and Bob traveled to Paris last summer to visit Microsoft headquarters."
annotations = proper_noun_annotator(sample_text)

# Printing the formatted output
print(f"Input Text: {sample_text}\n")
print(f"{'Word':<15} | {'Label':<6} | {'Character Span':<14}")
print("-" * 43)
for item in annotations:
    span = f"({item['start_char']}, {item['end_char']})"
    print(f"{item['text']:<15} | {item['label']:<6} | {span:<14}")
```
